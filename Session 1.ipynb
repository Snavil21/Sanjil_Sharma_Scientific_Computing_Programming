```python
#Solution 1#
def string_count(a):
    count = 0
    vowels = ['a', 'e', 'i', 'o', 'u']
    arr = list(a)
    for i in arr:
        if i in vowels:
            count += 1
    return count

a = input(str())
print(string_count(a))
```

```python
#Solution 2#
def string_count_2(a):
    keys = ['a', 'e', 'i', 'o', 'u']
    current = [0,0,0,0,0]
    b = dict(vowel = keys, count = current)
    arr = list(a)
    for i in keys:
        for j in arr:
            if i == j:
                current[i] += 1
    return b

a = input(str())
print(string_count_2(a))
```

```python
#Soltuion 3#
import math
def prime(a):
    if a <= 1:
        print('Not Prime')
    elif a == 2:
        print('Prime')
    elif a > 2 and a % 2 == 0:
        print('Not Prime')
    else:
        for i in range(3, int(math.sqrt(a))+1, 2):
            if a % i == 0:
                print('Not Prime')
        print('Prime')

a = int(input())
prime(a)
```

```python
#Solution 5#
def pattern(a):
    for i in range(1, a+1):
        for j in range(i):
            print(i, end=' ')
        print()
    return

a = int(input())
pattern(a)
```

```python
#Solution 6#
def pal(a):
    arr = list(a)
    i = 0
    j = len(arr)-1
    while i < j:
        if arr[i] == arr[j]:
            i += 1
            j -= 1
        else:
            return 'Not Palindrome'
    return 'Palindrome'

a = input()
print(pal(a))
```

```python
#Solution 7#
def rev(a):
    arr = list(a)
    i = 0
    j = len(arr)-1
    while i < j:
        arr[i], arr[j] = arr[j], arr[i]
        i += 1
        j -= 1
    return "".join(arr)

a = input()
print(rev(a))
```

```python
#Solution 9#
class operations:
    def __init__(self):
        self.data = []
    def sum(self, a, b):
        result = a+b
        self.data.append(f"{a} + {b} = {result}")
        return result
    def mul(self, a, b):
        result = a*b
        self.data.append(f"{a} * {b} = {result}")
        return result
    def div(self, a, b):
        if b == 0:
            return "Not Defined"
        else:
            result = a/b
            self.data.append(f"{a}/{b} = {result}")
            return result
    def sub(self, a, b):
        result = a-b
        self.data.append(f"{a} - {b} = {result}")
        return result

def calc(a,b,op):
    my_calc = operations()
    if op == 1:
        print(my_calc.sum(a, b))
    elif op == 2:
        print(my_calc.sub(a, b))
    elif op == 3:
        print(my_calc.mul(a, b))
    elif op == 4:
        print(my_calc.div(a, b))
    else:
        print('No Operations Selected')

op = int(input('Choose 1 for Add, 2 for Sub, 3 For Mul, 4 for Div'))
a = float(input())
b = float(input())
calc(a,b,op)
```

```python
#Solution 10#
def sort(a):
    b = [], c = []
    for i in a:
        if i%2 == 0:
            b.append(i)
        else:
            c.append(i)
    print('Even List:', b)
    print('Odd List:', c)

num = [int(i) for i in input().split(',')]
sort(num)
```

```python
#Solution 11#
def subject(a,b,c,d):
    marks = {
        'math':a,
        'english':b,
        'science':c,
        'computer':d
    }
    max_marks = max(marks, key=marks.get)
    print(max_marks)

a = int(input('Marks in Math:'))
b = int(input('Marks in English:'))
c = int(input('Marks in Science:'))
d = int(input('Marks in Computer:'))
subject(a,b,c,d)
```

```python
#Solution 12#
def duplicate(a):
    count = {}
    for i in a:
        if i in count:
            return True
        else:
            count[i] = 1
    return False

num = [int(i) for i in input().split(',')]
print(duplicate(num))
```

```python
#Solution 13#
def anagram(a,b):
    arr = list(a)
    brr = list(b)
    for i in arr:
        for j in brr:
            if i == j:
                arr.remove(i)
                brr.remove(j)
    if arr[0] == brr[0] and len(arr)==len(brr):
        print('Anagram')
    else:
        print('Not Anagram')

num_1 = input()
num_2 = input()
anagram(num_1, num_2)
```

```python
#Solution 14#
def stocks(arr):
    diff = []
    for i in range(len(arr)-1):
        for j in range(i+1, len(arr)):
            a = arr[j] - arr[i]
            diff.append(a)
    res = max(diff)
    if res <= 0:
        res = 0
    return res

prices = [int(i) for i in input().split(',')]
print(stocks(prices))
```

```python
#Solution 15#
def seq(a):
    nums_2 = set()
    final = 0
    for i in a:
        nums_2.add(i)
    for i in a:
        if i in nums_2 and i-1 not in nums_2:
            curr = i
            count = 0
            while curr in nums_2:
                count += 1
                curr += 1
            final = max(final, count)
    return final

nums = [int(i) for i in input().split(',')]
print(seq(nums))
```
